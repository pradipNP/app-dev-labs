```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("house_price_small_dataset.csv")
```

```python
# Extract feature value and target values
X = df["Size (sq ft)"]
y = df["Price ($)"]
```

```python
n = len(X)
```

```python
# Calculate slope and intercept for linear regression line
m = (n*np.sum(X*y) - np.sum(X)*np.sum(y)) / (n*np.sum(X**2) - np.sum(X)**2)
b = (np.sum(y) - m*np.sum(X)) / n
```

```python
print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
```

```python
# For prediction of price using the linear regression line
y_pred = m*X + b
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X, y, color="blue", label="Actual data") # we can use alpha = 0 to 1 for transparency
plt.plot(X, y_pred, color="red", label="Regression Line")
plt.title("Linear Regression: Size vs. Price")
plt.xlabel("Size (sq ft)")
plt.ylabel("Price ($)")
plt.legend()
plt.show()
```
