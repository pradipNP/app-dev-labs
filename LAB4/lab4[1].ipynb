```python
def second_largest(nums):
    nums = list(set(nums))
    nums.sort()
    return nums[-2]

my_list = [3, 1, 4, 1, 5, 9, 2, 6, 5]
print("Second Largest Number:", second_largest(my_list))
```

```python
def remove_duplicates(nums):
    return list(set(nums))

my_list = [1, 2, 2, 3, 4, 4, 5]
print("List without duplicates:", remove_duplicates(my_list))
```

```python
def tuple_to_string(tup):
    return ''.join(tup)

my_tuple = ('H', 'e', 'l', 'l', 'o')
print("Tuple to String:", tuple_to_string(my_tuple))
```

```python
def tuple_length(tup):
    return len(tup)

my_tuple = (1, 2, 3, 4, 5)
print("Length of Tuple:", tuple_length(my_tuple))
```

```python
def union_intersection(set1, set2):
    return set1.union(set2), set1.intersection(set2)

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
union, intersection = union_intersection(set1, set2)
print("Union:", union)
print("Intersection:", intersection)
```

```python
def is_subset(set1, set2):
    return set1.issubset(set2)

set1 = {1, 2}
set2 = {1, 2, 3, 4}
print("Is set1 a subset of set2?", is_subset(set1, set2))
```

```python
def count_frequency(my_list):
    freq = {}
    for item in my_list:
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1
    return freq

my_list = ['a', 'b', 'a', 'c', 'b', 'a']
print("Frequency of elements:", count_frequency(my_list))
```

```python
def merge_dictionaries(dict1, dict2):
    merged_dict = dict1.copy()
    merged_dict.update(dict2)
    return merged_dict

dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
print("Merged Dictionary:", merge_dictionaries(dict1, dict2))
```
